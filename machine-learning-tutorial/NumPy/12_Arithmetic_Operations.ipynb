## Matrix Operation with Scalar

```python
import numpy as np 
```

```python
#create 3*3 matrix 
m = np.array([[10,20,30], [5,15,25], [0,5,10]])
print("Matrix is created\n", m)
```

#### 1. Scalar Addition

```python
a = 10
print("Matrix is created\n ", m)
print("Scalar Addition \n", m+a)
```

#### 2. Scalar Subtraction

```python
a = 10
print("Matrix is created\n ", m)
print("Scalar Subtraction\n ", m-a)
```

#### 3. Scalar Division

```python
a = 3 
print("Matrix is created\n ", m)
print("Scalar Division\n ", m / a)
```

#### 4. Scalar Multiplication

```python
a = 10
print("Matrix is create\n", m)
print("Scalar Multiplication \n", m * a)
```

#### Element wise operations

Rule: 
* For matrix addition, subtraction, multiplication and division, both the matrix must be in same dimension (i.e) Same number of rows and columns

#### 1. Matrix Addition

```python
import numpy as np
```

```python
m1 = np.array([[10,20,30],[25,50,100],[20,40,60]])
m2 = np.array([[5,10,15],[1,50,100],[25,50,75]])
print("Matrix m1: \n", m1)
print("Matrix m2: \n", m2)
```

```python
print("Addition of two matrix = \n", m1 + m2)
```

```python
print("Addition of two matrix = \n", np.add(m1,m2))
```

#### 2. Matrix Subtraction 

```python
print("Matrix m1: \n", m1)
print("Matrix m2: \n", m2)
print("Subtraction of two matrix = \n", m1-m2)
```

```python
print("Subtraction of two matrix = \n", np.subtract(m1,m2))
```

#### 3. Matrix Division

```python
print("Matrix m1: \n", m1)
print("Matrix m2: \n", m2)
print("Division of two matrix = \n", m1/m2)
```

```python
print("Division of two matrix = \n", np.divide(m1,m2))
```

#### 4. Matrix Multiplication

```python
print("Matrix m1: \n", m1)
print("Matrix m2: \n", m2)
print("Multiplication of two matrix = \n", m1 * m2)
```

```python
print("Multiplication of two matrix = \n ", np.multiply(m1,m2))
```
