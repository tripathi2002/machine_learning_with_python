#### 9. argmin
numpy.argmin(array, axis=None, out=None) : Returns indices of the min element of the array in the particular axis.

```python
array = np.random.randint(16, size=(4,4))
print("INPUT ARRAY : \n",array)

# axis = 0 means its y axis (Column wise)
print("Indices of min element in y axis: ", np.argmin(array, axis=0))
# axis = 1 means its x axis (Row wise)
print("Indices of min element in x axis: ", np.argmin(array, axis=1))
```

#### 12. sum(axis)
Returns sum of the element of the array in a particular axis. 

```python
array = np.random.randint(20, size=(4,4))
print("INPUT ARRAY : \n", array)

# axis = 0 means its y axis 
print("Sum of Array Elements in y axis: ", np.sum(array, axis=0))

# axis = 1 means its x axis 
print("Sum of Array Elements in x axis: ", np.sum(array, axis=1))
```
