```python
# importing numpy 
import numpy as np
```

**2. Basic Slicing**

```python
# Creation Numpy Arrays
a = np.arange(10)
print(a)
```

```python
# Startindex: Endindex(exclusive)
# Slicing the elements from index 1 to 5
print(a[1:5])
```

```python
# Startindex: Endindex(exclusive) : Step
# Slicing the element using Step
print(a[1:9:2])
```

```python
# Alternative Ways of slicing with default values 

# End index is not mentioned
print(a[1:])

# Start index is not mentioned 
print(a[:9])
```

```python
# Assigning and slicing the elements 

# Array creation 
b = np.arange(10)
print(b)

# Assign the value 
b[5:] = 20
print(b)

# Slicing 
print(b[3:7])
```

```python
a = np.array([[1,2,3],[3,4,5],[4,5,6]])
print(a)

# Slice all items from the second row 
print(a[1, ...])
# Slice all items from second column 
print(a[...,1])
```
