### 1. Changing Shape of Array 
***

#### 1. reshape()
Gives a new shape to an array without changing its data.

Syntax: 
__numpy.reshape(arr,newshape, order')__

Note: 
* New shape must be compatible to the original shape
* Order ='C': Its arrange the value in the row wise (Default Value)
* Order ='F': Its arrange the value in the column wise 

# Execute this code for help or additional examples 
help()

```python
help(np.shape)
```

```python
# Generating 9 numbers using arange function 
a = np.arange(9)
print(a)
```

```python
# Checking the shape and dimension
print(a.ndim)
print(a.shape)
print(a.size)
```

```python
# Reshaping the array 'a' to 3*3 matrix 
b = np.reshape(a, (3,3))
print(b)
```

```python
# reshaping the array 'a' to 3*3 matrix with order as 'F'
b = np.reshape(a, (3,3), order='F')
print(b)
```

```python
# Checking the shape and dimension
print(b.ndim)
print(b.shape)
print(b.size)
```

***
Note: Have to metion the shape exactly matching the size of the array
***

```python
# Negative Scenario 
c = np.reshape(a, (2,2))
print(c)
```

```python
# Negative Scenario 
c = np.reshape(a, (3,4))
print(c)
```

#### 2. ravel()
This function returns a flattened one-dimensional array. The returned array will have the same type as that of the input array.

The function takes one parameter.

Syntax: 
**numpy.ravel(a, order)**

Note:
* Order ='C': Its arrange the value in the row wise (Default Value)
* Order ='F': Its arrange the value in the column wise

```python
# creating an array using arange and reshape function 
a = np.arange(12).reshape(3,4)
print(a)
```

```python
# Checking the shape and dimension 
print(a.ndim)
print(a.shape)
print(a.size)
```

```python
# Implementiong the ravel method - To convert array to 1D 
b = np.ravel(a, order="C")
print(b)
b = np.ravel(a, order="F")
print(b)
```

```python
# Checking the shape and dimension 
print(b.ndim)
print(b.shape)
print(b.size)
```

### 2. Transpose Operations 
***
#### 1. transpose()
This function changes shape of the array by swapping the position with the row by columns and viz

Syntax: 
__numpy.transpose(arr, axes)__


```python
# Creating the array using reshape method 
a = np.arange(10).reshape(5,2)
print(a)
print(a.shape)
```

```python
# Transpose of a metrix
b = np.transpose(a)
print(b)
print(b.shape)
```

```python
# Another way to do transpose with Letter 'T'
c = a.T
print(c)
```

#### swapaxes()
This function interchanges the two axes of an array 

Syntax: 

__numpy.swapaxes(arr, axis1, axis2)__

```python
# Creating an array 
a = np.arange(12).reshape(4,3)
print(a)
```

```python
b = np.swapaxes(a,0,1)
print(b)
```

### 3. Adding / Removing Elements

#### 1. resize()

* This function returns a new array with the specified size.

it repeats the original array till the metioned limit the repeated copies of entries in the original are contained. 

Syntax: 

__numpy.resize(arr,shape)__

Where, 
* arr - Input Array
* shape - Shope of an array which need to be created 

__Array Creation__

```python
# Creating array using arange function 
a = np.arange(12)
print(a)
```

__Example 1: Size of new array equal than original array size__

```python
# Resizing the array as (3*4) shape 
b = np.resize(a,(3,4))
print(b)
```

__Example 2: Size of new array is less than original array size__

```python
# Resizing the array as (3*3) shape - size of array less than oringinal array 
c = np.resize(a,(3,3))
print(c)
```

__Example 3: Size of new array is greater than oringinal array size__

```python
# Resizing the array as (4*4) shape - size of array more than oringinal array 
d = np.resize(a,(4,4))
print(d)
```

#### 2. append() 
* This function append values at the end of an input array.
* The dimensions of the append values must match with input array otherwise ValueError will be generated.
* If the axis is not mentioned, then 1D array will be created by appending values at the end

Syntax: 

__numpy.append(arr,values,axis)__

Where, 
* arr - Exisiting Array
* values -Values to append. Must match the existing shape of array
* axis - The axis along which to insert. If not given, the input array is flattened.

```python
# Array Crreation - Creating an array 
a = np.array([[1,2,3],[4,5,6]])
print(a)
```

__Example 1: Appending elements without axis parameter__

```python
# Append Elements to the array - If axis not mentioned, it will append and create 1D array
np.append(a,[[10,10,10]])
```

__Example 2: Appending element row wise__

```python
print(a)
# Appending Elements to the array - Adding 2D values in row wise(axis = 0)
np.append(a, [[10,10,10]],axis=0)
```

__Example 3: Appending element column wise__

```python
print(a)
# Appending Element to the array - Adding 2D values in column wise(axis=1)
np.append(a,[[10,10], [30,20]], axis=1)
```

#### 3. insert() 
* This function inserts values in the input array along the given axis and before the given index.
* If Insertion is not done in place and the funciton return a new array

Syntax: 

__numpy.insert(arr, obj, values, axis)__

Where, 
* arr - Input Array
* obj - The index before which insertion is to be made
* values - Values to append. Must match the existing shape of array
* axis - The axis along which to insert. If not given, the input array is flattened

```python
# Array Creation - Creating an Array 
a = np.array([[1,2,3],[4,5,6]])
print(a)
```

__Example 1: Inserting value without axis parameter at index 1( 1D array as output)__

```python
np.insert(a,1,[10,11,12])
```

__Example 2: Inserting value along axis 0 at index 1__

```python
np.insert(a,1,[10,11,12],axis=0)
```

__Example 3: Inserting value along axis 0 at index 1 using single value(broadcasting)__

```python
np.insert(a,1,[20],axis=0)
```

__Example 4: Inserting value along axis 1 at index 1__

```python
print(a)
np.insert(a,1,[10,11],axis=1)
```

__Example 5: Inserting value along axis 1 at index 1 using single value(broadcasting)__

```python
np.insert(a,1,[10],axis=1)
```

#### 4. delete()
* This function returns a new array with the specified subarray deleted from the input array

Syntax:

__numpy.delete(arr,obj,axis)__

Where, 
* arr - Input Array
* obj - Can be a slice, an integer or array of integers, indicating the subarray to be deleted from the input array
* axis - The axis along which to delete the given subarray. If not given, arr is flattened

```python
# Array Creation - Creating an array 
a = np.arange(12).reshape(4,3)
print(a)
```

__Example 1: Deleting value without axis parameter at index 2 (1D array as output)__

```python
np.delete(a,2)
```

__Example 2: Deleting value along axis 0 at index 2__

```python
np.delete(a,2,axis=0)
```

Example 3: Deleting value along axis 1 at index 2 

```python
print(a)
np.delete(a,2,axis=1)
```

__Example 4: Deleting values using slicing__

```python
# Creating the array
a = np.arange(12)
print(a)
```

```python
# Slicing using starting and ending index 
print(np.delete(a,np.s_[3:7]))
```

```python
print(a)
# Slicing using step size 2 
print(np.delete(a,np.s_[::2]))
```
