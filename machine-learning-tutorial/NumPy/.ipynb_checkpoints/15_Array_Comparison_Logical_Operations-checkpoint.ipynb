### Array Comparison 

```python
import numpy as np
```

```python
# Creating 2 arrays to compare element wise
a = np.array([1,2,3,4,5])
b = np.array([2,3,5,4,1])
```

#### Element Wise Comparison

```python
print(a==b)
```

```python
print(a>b)
```

```python
print(a<b)
```

#### Array Wise Comparison

```python
a = np.array([1,2,3,4])
b = np.array([2,4,6,8])
c = np.array([1,2,3,4])
```

```python
np.array_equal(a,b)
```

```python
np.array_equal(a,c)
```

### Logical Operations

```python
# Creating 2 array's 
a = np.array([1,1,0,0])
b = np.array([1,0,1,0])
```

#### logical_or()

```python
np.logical_or(a,b)
```

#### logical_and()

```python
np.logical_and(a,b)
```

#### all()

```python
np.all([True, True, False, True])
```

```python
a = np.array([1,2,3,4])
b = np.array([2,4,6,3])
c = np.array([1,2,3,4])
((a<=b) & (a<=c))
```
