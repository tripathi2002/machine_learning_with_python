# 1. Array Creation using Ones and Zeros 

1. numpy.empty

   Return a new array of given shape and type, without initializing enteries.

```python
#Code to create empty array 
import numpy as np
np.empty([2,2])
```

2. numpy.zeros

   Return a new array of given shape and type, filled with zeros

```python
#array of three zeros. Default dtype is float 
a = np.zeros(3)
print(a)
```

```python
b = np.zeros((3,3))
print(b)
```

3. numpy.ones

   Return a new array of given shape and type, filled with ones 

```python
a = np.ones(3)
print(a)
```

```python
b = np.ones((3,3,))
print(b)
```

4. numpy.eye

   Return a 2-D array with ones on the diagonal and zeros elsewhere. It is also called as Diagonal matrix.

```python
a = np.eye(2)
print(a)
```

```python
b = np.eye(3)
print(b)
```

5. numpy.full

   Return a new array of given shape and type, filled with fill_value. 

```python
#Code to create array as per mentioned shape and value 
a = np.full((2,4), 5)
print(a)
```

```python
b = np.full((3,2), 10)
print(b)
```
