```python
#Importing the NumPy library 
import numpy as np
```

1 Dimensional Numpy Array 

```python
#1-Dimensional Array - Using List 
a = np.array([1.,2.,3.,4.,5.])
print(a)
```

2 Dimensional Numpy Array

```python
# 2-Dimensional Array 
b = np.array([[1,2,3],[4,5,6]])
print(b)
```

3 Dimensional Array

```python
# 3-Dimensional Array
c = np.array([[[1,2],[3,4]],[[5,6],[7,8]]])
print(c)
```

# Lecture 3 - Array Attributes for 1D, 2D and 3D for above examples 

1. ndim - It returns the number of array dimensions 

```python
print(a.ndim)
print(b.ndim)
print(c.ndim)
```

2. shape - It returns a tuple consisting of array dimensions. It can also be used to resize the array. 

```python
print(a.shape)
print(b.shape)
print(c.shape)
```

Note: We will learn Reshapes concept in later chapters

3. len - It returns length of the array. For ndimension it returns the size of the first dimension  

print(len(a))
print(len(b))
print(len(c))


4. dtype - It return the data type of the array.

```python
print(a.dtype)
print(b.dtype)
print(c.dtype)
```

5. size - It returns the total number of the elements inside the array.

```python
print(a.size)
print(b.size)
print(c.size)
```
