### Boolean Indexing

```python
import numpy as np
```

```python
c = np.arange(10,100,5)
c
```

```python
#Boolean Condition for each elements of array
c>49
```

```python
#Display the values which satisfy the boolean condition
c[c>49]
```

Logical And

```python
#Array Creation 
num = np.arange(0,101,10)
print(num)

#Logical And conditions
Bgrade = np.logical_and(num>35, num<=60)

#Display the values which satisfy the boolean condition 
num[Bgrade]
```

Where Condition

```python
#Where Condition 
# val = np.where(num>30, num, 0)  # where(condition, array_list, replace_value)
val = np.where(num>30, num, 'NA') 

#replace the element with mentioned values for false conditions
val
```
