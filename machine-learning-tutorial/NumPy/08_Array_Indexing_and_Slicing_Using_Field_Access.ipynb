```python
#importing numpy
import numpy as np
```

**1. Field access**

```python
# Accessing an Element 
a = np.arange(10)

print("Numpy array is ", a)
print("Element at mentioned index is ", a[5])
```

```python
# start, end (exclusive), step
b = np.arange(10,30,2)

print("Numpy array is ", b)
print("Element at mentioned index is ", b[5])
```

```python
# For multidimensional arrays, indexes are tuples of integers 
c = np.array([[1,2,3],[4,5,6]])

#print the array 
print(c)

# Note: first array/column starts at 0. 
# Accessing elements using row, column index
print(c[0,0])
print(c[0,2])
```

```python
# Assigning or Updating the value
print(c)
c[0,0] = 10
print(c)

c[1,2] = 25
print(c)
```
