# Array Creation using Existing Data

```python
# importing numpy library
import numpy as np
```

1. numpy.asarray

   It Convert the input to an array. It is useful for converting Python sequence into ndarray. 

```python
# convert list to ndarray
lst = [1,2,3,4,5]

np.asarray(lst)
```

```python
# convert tuple to ndarray
tple = (1,2,3,4,5)

np.asarray(tple)
```

```python
# ndarray from list of tuples
x = [(1,2,3),(4,5,6)]

np.asarray(x)
```

```python
# dtype is set as float 
x = [(1,2,3),(4,5,6)]

np.array(x, dtype = float)
```

2. numpy.copy

   Return an array copy of the given object. 

```python
#Create an array a and copy to b: 
a = np.asarray([1,2,3,4,5])
b = np.copy(a)

print("Original Array", a)
print("Copy of an Array", b)
```

3. numpy.frombuffer

   Interpret a buffer as a 1-dimensional array. 

```python
s = b'Hello'
np.frombuffer(s, dtype = 'S1')
```

4. numpy.fromiter

   Create a new 1-dimensional array from an iterable object.

```python
# code to create array using list 
iterable = range(10)
np.fromiter(iterable, int)
```

```python
# code to create array for square of mentioned range 
iterable = (x*x for x in range(10))
np.fromiter(iterable, float)
```
